```python
from astropy.io import fits
import matplotlib.pyplot as plt
from astropy.visualization import ImageNormalize, AsinhStretch
import numpy as np
from vis import plot_model_residual, plot_data_psf
import numpy as np
from scipy.ndimage import map_coordinates, zoom
from inspect_target import data_info
from plot_flux import plot_flux_slice
```

## Case 1: Filter = F356W

```python
## here my science cutout is at 0.05”/pix, while the PSF you want to use is on 0.063”/pixel.

data = fits.open("../data/J0148-12884/F356W.fits")
psf = fits.open("../PSF_files/J0148-12884/F356W.fits")


#print(psf[1].header.get("PIXELSCL"))
print(data[0].header.get("CD1_1"))             # pixel_scale = abs(-1.3888889e-05) * 3600 ≈ 0.05 arcsec/pixel
data_info('../data/J0148-12884/F356W.fits')
```

```python
data = fits.getdata("../data/J0148-12884/F356W.fits")
psf = fits.getdata("../PSF_files/J0148-12884/F356W.fits")
plot_data_psf(data[90:110, 90:110], psf[90:110, 90:110], cmap="RdBu_r")

residual = data - psf 
v = np.nanmax(np.abs(residual))
plt.imshow(residual[90:110, 90:110], origin="lower", cmap="RdBu_r")
plt.colorbar(label="Data − scaled PSF")
plt.show()
```

```python
##trying 2 sersic
data = fits.getdata("../data/J0148-12884/F356W.fits")
model = fits.getdata("../imfit_models/J0148-12884/F356W_2sersic.fits")
residual = fits.getdata("../imfit_resid/J0148-12884/F356W_2sersic.fits")

plot_model_residual(data[90:110, 90:110], model, residual, model_title="Two-Sersic Model", cmap="RdBu_r")
plot_flux_slice(data[90:110, 90:110], model, residual, 8.5, 0, 20)
```
