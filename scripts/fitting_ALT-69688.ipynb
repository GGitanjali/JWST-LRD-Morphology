```python
from astropy.io import fits
import matplotlib.pyplot as plt
from astropy.visualization import ImageNormalize, AsinhStretch
import numpy as np
from vis import plot_model_residual, plot_data_psf
from scipy.ndimage import map_coordinates, zoom
from inspect_target import data_info
from plot_flux import plot_flux_slice
```

## Case 1: Filter = F356W

```python
## here my science cutout is at 0.05”/pix, while the PSF you want to use is on 0.063”/pixel.

data = fits.open("../data/ALT-69688/F356W.fits")
psf = fits.open("../PSF_files/ALT-69688/F356W.fits")


#print(psf[1].header.get("PIXELSCL"))
print(data[0].header.get("CD1_1"))             # pixel_scale = abs(-1.3888889e-05) * 3600 ≈ 0.05 arcsec/pixel
data_info('../data/ALT-69688/F356W.fits')
```

```python
data = fits.getdata("../data/ALT-69688/F356W.fits")
psf = fits.getdata("../PSF_files/ALT-69688/F356W.fits")
data_new = data[90:110, 90:110]
psf_new = psf[90:110, 90:110]
plot_data_psf(data_new, psf_new, cmap="RdBu_r")

## data - psf

residual = data - psf 
v = np.nanmax(np.abs(residual))
plt.imshow(residual[91:110,91:110], origin="lower", cmap="RdBu_r")
plt.colorbar(label="Data − scaled PSF")
plt.show()
```

```python
data = fits.getdata("../data/ALT-69688/F356W.fits")
model = fits.getdata("../imfit_models/ALT-69688/F356W_sersic.fits")
residual = fits.getdata("../imfit_resid/ALT-69688/F356W_sersic.fits")

plot_model_residual(data[90:110, 90:110], model, residual, model_title="Sersic Model", cmap="RdBu_r")
plot_flux_slice(data[90:110, 90:110], model, residual, 8.5, 0, 20)
```
