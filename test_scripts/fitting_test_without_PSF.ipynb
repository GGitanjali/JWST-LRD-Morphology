```python
from vis import plot_model_residual
from astropy.io import fits 
```

```python
data = fits.getdata("../data/F277W.fits")
model = fits.getdata("../imfit_models/F277W_model.fits")
residual = fits.getdata("../imfit_resid/F277W_resid.fits")

plot_model_residual(data, model, residual, model_title="Gaussian Model", cmap="RdBu_r", xlim=(50, 150), ylim=(50, 150))
```
